```python
"""
Year 2 General – Gender Distribution Charts
============================================
Generates three chart images and saves them to a folder called "charts/"
in the same directory as this script.

Requirements:
    pip install matplotlib

Run:
    python gender_charts.py
"""

import os
import matplotlib
matplotlib.use('Agg')          # non-interactive backend (safe for all platforms)
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# ── Data ─────────────────────────────────────────────────────────────────────
BOYS   = 179
GIRLS  = 82
TOTAL  = BOYS + GIRLS          # 261

# ── Colours ──────────────────────────────────────────────────────────────────
BOY_COLOR  = '#185fa5'         # blue
GIRL_COLOR = '#b03a6e'         # pink/rose
BG_COLOR   = '#F8F9FC'         # very light grey-blue background

# ── Output folder ─────────────────────────────────────────────────────────────
OUT_DIR = os.path.join(os.path.dirname(os.path.abspath(__file__)), 'charts')
os.makedirs(OUT_DIR, exist_ok=True)

# ── Shared matplotlib style ───────────────────────────────────────────────────
plt.rcParams.update({
    'font.family':        'DejaVu Sans',
    'axes.spines.top':    False,
    'axes.spines.right':  False,
    'figure.facecolor':   BG_COLOR,
    'axes.facecolor':     BG_COLOR,
})


# ══════════════════════════════════════════════════════════════════════════════
# Chart 1 — Doughnut chart
# ══════════════════════════════════════════════════════════════════════════════
fig, ax = plt.subplots(figsize=(6, 5))
fig.patch.set_facecolor(BG_COLOR)
ax.set_facecolor(BG_COLOR)

wedges, _ = ax.pie(
    [BOYS, GIRLS],
    colors=[BOY_COLOR, GIRL_COLOR],
    startangle=90,
    wedgeprops=dict(width=0.52, edgecolor='white', linewidth=3),
)

# Centre label showing total
ax.text(0,  0.12, str(TOTAL),    ha='center', va='center',
        fontsize=26, fontweight='bold', color='#1a1a2e')
ax.text(0, -0.20, 'students',    ha='center', va='center',
        fontsize=11, color='#555')

# Legend
ax.legend(
    handles=[
        mpatches.Patch(facecolor=BOY_COLOR,
                       label=f'Boys  —  {BOYS}  ({BOYS / TOTAL * 100:.1f}%)'),
        mpatches.Patch(facecolor=GIRL_COLOR,
                       label=f'Girls  —  {GIRLS}  ({GIRLS / TOTAL * 100:.1f}%)'),
    ],
    loc='lower center', bbox_to_anchor=(0.5, -0.07),
    ncol=2, frameon=False, fontsize=11,
)

ax.set_title('Gender Distribution — Year 2 General',
             fontsize=13, fontweight='bold', color='#1a1a2e', pad=14)

plt.tight_layout()
path1 = os.path.join(OUT_DIR, 'chart1_doughnut.png')
plt.savefig(path1, dpi=150, bbox_inches='tight', facecolor=BG_COLOR)
plt.close()
print(f'Saved: {path1}')


# ══════════════════════════════════════════════════════════════════════════════
# Chart 2 — Vertical bar chart
# ══════════════════════════════════════════════════════════════════════════════
fig, ax = plt.subplots(figsize=(6, 5))
fig.patch.set_facecolor(BG_COLOR)
ax.set_facecolor(BG_COLOR)

bars = ax.bar(
    ['Boys', 'Girls'],
    [BOYS, GIRLS],
    color=[BOY_COLOR, GIRL_COLOR],
    width=0.45,
    edgecolor='white',
    linewidth=0,
    zorder=3,
)

# Value + percentage labels above each bar
for bar, val in zip(bars, [BOYS, GIRLS]):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 3,
        f'{val}\n({val / TOTAL * 100:.1f}%)',
        ha='center', va='bottom',
        fontsize=12, fontweight='bold', color='#1a1a2e',
    )

ax.set_ylim(0, 230)
ax.set_ylabel('Number of students', fontsize=11, color='#444')
ax.set_title('Boys vs Girls — Year 2 General',
             fontsize=13, fontweight='bold', color='#1a1a2e', pad=14)
ax.yaxis.grid(True, linestyle='--', alpha=0.5, zorder=0)
ax.set_axisbelow(True)
ax.tick_params(labelsize=12)
ax.spines['left'].set_color('#ccc')
ax.spines['bottom'].set_color('#ccc')

plt.tight_layout()
path2 = os.path.join(OUT_DIR, 'chart2_bar.png')
plt.savefig(path2, dpi=150, bbox_inches='tight', facecolor=BG_COLOR)
plt.close()
print(f'Saved: {path2}')


# ══════════════════════════════════════════════════════════════════════════════
# Chart 3 — Horizontal stacked proportion bar
# ══════════════════════════════════════════════════════════════════════════════
fig, ax = plt.subplots(figsize=(7.5, 2.2))
fig.patch.set_facecolor(BG_COLOR)
ax.set_facecolor(BG_COLOR)

b_pct = BOYS  / TOTAL * 100   # 68.6 %
g_pct = GIRLS / TOTAL * 100   # 31.4 %

ax.barh(['Year 2 General'], [b_pct],
        color=BOY_COLOR,  height=0.45, label='Boys')
ax.barh(['Year 2 General'], [g_pct],
        color=GIRL_COLOR, height=0.45, left=[b_pct], label='Girls')

# Labels inside the bars
ax.text(b_pct / 2,        0, f'Boys\n{BOYS} ({b_pct:.1f}%)',
        ha='center', va='center', fontsize=11, fontweight='bold', color='white')
ax.text(b_pct + g_pct / 2, 0, f'Girls\n{GIRLS} ({g_pct:.1f}%)',
        ha='center', va='center', fontsize=11, fontweight='bold', color='white')

ax.set_xlim(0, 100)
ax.set_xlabel('Percentage (%)', fontsize=11, color='#444')
ax.set_title('Gender Proportion — Year 2 General',
             fontsize=13, fontweight='bold', color='#1a1a2e', pad=12)
ax.xaxis.grid(True, linestyle='--', alpha=0.4, zorder=0)
ax.set_axisbelow(True)
ax.spines['left'].set_color('#ccc')
ax.spines['bottom'].set_color('#ccc')
ax.tick_params(axis='y', labelsize=11)

plt.tight_layout()
path3 = os.path.join(OUT_DIR, 'chart3_hbar.png')
plt.savefig(path3, dpi=150, bbox_inches='tight', facecolor=BG_COLOR)
plt.close()
print(f'Saved: {path3}')

print('\nAll 3 charts saved to the "charts/" folder.')
```
