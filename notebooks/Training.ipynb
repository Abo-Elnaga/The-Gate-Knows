```python
import os
import dlib
import csv
import numpy as np
import logging
import cv2

# Path of main folder (contains subfolders of persons)
path_images_from_camera = r"D:\AI\Projects\AI_Final_Project\Year_two\Updated Data"

# Dlib models
detector = dlib.get_frontal_face_detector()
predictor = dlib.shape_predictor('shape_predictor_68_face_landmarks.dat')
face_reco_model = dlib.face_recognition_model_v1("dlib_face_recognition_resnet_model_v1.dat")


# Return 128D features for single image
def return_128d_features(path_img):
    img_rd = cv2.imread(path_img)

    if img_rd is None:
        logging.warning(f"Cannot read image: {path_img}")
        return 0

    faces = detector(img_rd, 1)

    logging.info(f"Image: {path_img} - Faces detected: {len(faces)}")

    if len(faces) != 0:
        shape = predictor(img_rd, faces[0])
        face_descriptor = face_reco_model.compute_face_descriptor(img_rd, shape)
        return np.array(face_descriptor)
    else:
        logging.warning("No face detected")
        return 0


# Return mean features for a person (including subfolders)
def return_features_mean_personX(path_face_personX):
    features_list_personX = []

    for root, dirs, files in os.walk(path_face_personX):
        for file in files:
            img_path = os.path.join(root, file)

            # skip non-image files
            if not file.lower().endswith(('.png', '.jpg', '.jpeg')):
                continue

            logging.info(f"Reading image: {img_path}")

            features_128d = return_128d_features(img_path)

            if isinstance(features_128d, np.ndarray):
                features_list_personX.append(features_128d)

    if features_list_personX:
        features_mean_personX = np.mean(features_list_personX, axis=0)
    else:
        features_mean_personX = np.zeros(128)

    return features_mean_personX


def main():
    logging.basicConfig(level=logging.INFO)

    # Get all person folders
    person_list = os.listdir(path_images_from_camera)
    person_list.sort()

    with open(r"C:\Users\FreeComp\Face-Recognition-Based-Attendance-System-main\All_Features.csv", "w", newline="") as csvfile:
        writer = csv.writer(csvfile)

        for person in person_list:
            person_path = os.path.join(path_images_from_camera, person)

            if not os.path.isdir(person_path):
                continue

            logging.info(f"Processing person: {person}")

            # Get mean features
            features_mean_personX = return_features_mean_personX(person_path)

            # Extract person name
            if len(person.split('_', 2)) == 2:
                person_name = person
            else:
                person_name = "_".join(person.split("_")[1:])

            # Add name at beginning
            row = [person_name] + list(features_mean_personX)

            writer.writerow(row)
            logging.info("Saved\n")

    logging.info("All features saved to: features_all.csv")


if __name__ == '__main__':
    main()
```

```python
import os
import dlib
import csv
import numpy as np
import logging
import cv2

# Paths
path_images_from_camera = r"D:\AI\Projects\AI_Final_Project\Year_two\Updated Data"
output_csv = r"C:\Users\FreeComp\Face-Recognition-Based-Attendance-System-main\All_Features.csv"
progress_file = r"C:\Users\FreeComp\Face-Recognition-Based-Attendance-System-main\progress.txt"

# Dlib models
detector = dlib.get_frontal_face_detector()
predictor = dlib.shape_predictor('shape_predictor_68_face_landmarks.dat')
face_reco_model = dlib.face_recognition_model_v1("dlib_face_recognition_resnet_model_v1.dat")


def return_128d_features(path_img):
    img = cv2.imread(path_img)
    if img is None:
        return None

    faces = detector(img, 1)
    if len(faces) == 0:
        return None

    shape = predictor(img, faces[0])
    face_descriptor = face_reco_model.compute_face_descriptor(img, shape)
    return np.array(face_descriptor)


def return_features_mean_person(path):
    features = []

    for root, dirs, files in os.walk(path):
        for file in files:
            if not file.lower().endswith(('.jpg', '.jpeg', '.png')):
                continue

            img_path = os.path.join(root, file)
            f = return_128d_features(img_path)

            if isinstance(f, np.ndarray):
                features.append(f)

    if len(features) > 0:
        return np.mean(features, axis=0)
    return np.zeros(128)


# Load last processed person
def load_last_progress():
    if os.path.exists(progress_file):
        with open(progress_file, "r") as f:
            return f.read().strip()
    return None


# Save progress
def save_progress(person):
    with open(progress_file, "w") as f:
        f.write(person)


def load_existing_people():
    existing = set()

    if os.path.exists(output_csv):
        with open(output_csv, "r", newline="", encoding="utf-8") as f:
            reader = csv.reader(f)
            for row in reader:
                if len(row) > 0:
                    existing.add(row[0])

    return existing


def main():
    logging.basicConfig(level=logging.INFO)

    person_list = sorted(os.listdir(path_images_from_camera))

    existing_people = load_existing_people()
    last_done = load_last_progress()

    # skip until last completed
    start_processing = True if last_done is None else False

    with open(output_csv, "a", newline="", encoding="utf-8") as csvfile:
        writer = csv.writer(csvfile)

        for person in person_list:
            person_path = os.path.join(path_images_from_camera, person)

            if not os.path.isdir(person_path):
                continue

            # skip already in CSV
            if person in existing_people:
                logging.info(f"Skip already encoded: {person}")
                continue

            # resume logic
            if not start_processing:
                if person == last_done:
                    start_processing = True
                continue

            logging.info(f"Processing: {person}")

            features = return_features_mean_person(person_path)

            writer.writerow([person] + list(features))
            csvfile.flush()  # مهم جدًا عشان مايضيعش الداتا

            save_progress(person)
            logging.info(f"Saved + checkpoint: {person}")

    logging.info("DONE")


if __name__ == "__main__":
    main()
```
